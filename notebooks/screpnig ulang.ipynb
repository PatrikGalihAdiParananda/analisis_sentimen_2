```python
# 1. Import library yang dibutuhkan
# google-play-scraper untuk mengambil data dari Play Store
# pandas untuk mengolah data dan menyimpannya ke CSV
from google_play_scraper import Sort, reviews
import pandas as pd

# 2. Proses scraping data ulasan dari Google Play Store
print("Memulai proses scraping ulasan...")

# Melakukan request ke Google Play untuk mendapatkan ulasan
# continuation_token digunakan jika kita ingin melanjutkan scraping dari sesi sebelumnya
result, continuation_token = reviews(
    'app.bpjs.mobile',
    lang='id',                               # Bahasa ulasan yang diambil adalah Bahasa Indonesia
    country='id',                            # Wilayah Google Play Store adalah Indonesia
    sort=Sort.NEWEST,                        # Mengurutkan ulasan dari yang paling baru
    count=5000,                              # Jumlah MAKSIMAL ulasan yang ingin diambil
    filter_score_with=3                      # INI BAGIAN PENTING: Hanya mengambil ulasan dengan rating bintang 3
)

# Jika tidak ada ulasan yang ditemukan, hentikan program
if not result:
    print("Tidak ada ulasan bintang 3 yang ditemukan atau terjadi kesalahan.")
else:
    # 3. Menyimpan data ke dalam file CSV
    # Mengubah hasil scraping (yang berupa list of dictionary) menjadi DataFrame pandas
    df_reviews = pd.DataFrame(result)
    
    # Menentukan nama file output
    nama_file_csv = 'ulasan_bintang_3_bpjs_mobile.csv'
    
    # Menyimpan DataFrame ke file CSV
    # index=False agar nomor index dari pandas tidak ikut disimpan dalam file CSV
    df_reviews.to_csv(nama_file_csv, index=False)
    
    print(f"Scraping selesai!")
    print(f"Sebanyak {len(df_reviews)} ulasan telah berhasil disimpan di file '{nama_file_csv}'")
```

```python
import pandas as pd
import joblib
import numpy as np

# --- Fungsi Heuristik untuk Menemukan Ketidakcocokan Label ---
def flag_heuristic_errors(df):
    """Menandai baris yang kemungkinan besar salah label berdasarkan aturan kata kunci."""
    
    # Kata kunci yang sangat kuat indikasinya
    strong_negative_words = ['kecewa', 'parah', 'buruk', 'error', 'lambat', 'lemot', 'lelet', 'sampah', 'bohong', 'uninstall']
    strong_positive_words = ['terima_kasih', 'terimakasih', 'sangat_membantu', 'luar_biasa', 'mantap', 'sempurna', 'mudah_sekali']
    
    flagged_indices = []
    
    for index, row in df.iterrows():
        text = row['processed_text']
        label = row['Label']
        
        # Aturan 1: Teks mengandung kata negatif kuat, tapi labelnya Positif
        if any(word in text for word in strong_negative_words) and label == 'Positif':
            flagged_indices.append(index)
            
        # Aturan 2: Teks mengandung kata positif kuat, tapi labelnya Negatif
        elif any(word in text for word in strong_positive_words) and label == 'Negatif':
            flagged_indices.append(index)
            
    return df.loc[flagged_indices]


# --- ALUR KERJA UTAMA ---
if __name__ == "__main__":
    try:
        # 1. MEMUAT SEMUA ASET
        print("📥 1. Memuat dataset, model, dan vectorizer...")
        df = pd.read_csv("data_35k_cleaned.csv") # Ganti dengan nama file CSV Anda yang sudah bersih
        df.dropna(subset=['processed_text', 'Label'], inplace=True)
        df['processed_text'] = df['processed_text'].astype(str)
        
        model = joblib.load("model_rf_terbaik.pkl")
        vectorizer = joblib.load("tfidf_vectorizer_terbaik.pkl")
        print("✅ Aset berhasil dimuat.")

        # 2. TRANSFORMASI DATA & PREDIKSI PADA SELURUH DATASET
        print("\n🔄 2. Melakukan prediksi pada seluruh dataset untuk analisis...")
        X_tfidf_full = vectorizer.transform(df['processed_text'])
        
        # Dapatkan prediksi dan probabilitasnya
        predictions = model.predict(X_tfidf_full)
        probabilities = model.predict_proba(X_tfidf_full)
        
        df['prediksi_model'] = predictions
        # Dapatkan probabilitas dari kelas yang diprediksi
        df['probabilitas_prediksi'] = np.max(probabilities, axis=1)
        
        print("✅ Prediksi selesai.")

        # 3. MENERAPKAN STRATEGI UNTUK MENEMUKAN DATA MENCURIGAKAN
        print("\n🔎 3. Menerapkan strategi untuk menemukan label yang perlu diperiksa...")

        # Strategi 1: Keyakinan Rendah (Low Confidence)
        # Ambil data di mana probabilitas prediksi model di bawah 70%
        low_confidence_df = df[df['probabilitas_prediksi'] < 0.70].copy()
        
        # Strategi 2: Kesalahan dengan Keyakinan Tinggi (Confident Errors)
        # Cari data yang salah prediksi (Label != prediksi_model)
        misclassified_df = df[df['Label'] != df['prediksi_model']].copy()
        # Dari yang salah, cari yang keyakinan modelnya di atas 90% (paling fatal)
        confident_errors_df = misclassified_df[misclassified_df['probabilitas_prediksi'] > 0.90].copy()

        # Strategi 3: Aturan Heuristik
        heuristic_errors_df = flag_heuristic_errors(df)

        print(f"  - Ditemukan {len(low_confidence_df)} data dengan keyakinan rendah.")
        print(f"  - Ditemukan {len(confident_errors_df)} data yang salah prediksi dengan keyakinan tinggi.")
        print(f"  - Ditemukan {len(heuristic_errors_df)} data yang melanggar aturan heuristik.")

        # 4. MENYIMPAN HASIL UNTUK REVIEW MANUAL
        print("\n💾 4. Menyimpan data yang perlu direview ke dalam file CSV...")
        
        # Gabungkan semua kandidat dan hapus duplikat
        review_candidates = pd.concat([low_confidence_df, confident_errors_df, heuristic_errors_df]).drop_duplicates().sort_index()
        
        # Tambahkan kolom untuk koreksi manual Anda
        review_candidates['label_koreksi_manual'] = ''
        
        output_filename = 'DATA_UNTUK_DIREVIEW.csv'
        review_candidates.to_csv(output_filename, index=False)
        
        print(f"\n✅ SELESAI! Anda tidak perlu memeriksa 35,000 data.")
        print(f"Cukup periksa {len(review_candidates)} data yang paling mencurigakan di file '{output_filename}'.")
        print("Buka file tersebut di Excel/Google Sheets dan isi kolom 'label_koreksi_manual'.")

    except FileNotFoundError:
        print("❌ ERROR: Pastikan file 'data_35k_cleaned.csv', 'model_rf_terbaik.pkl', dan 'tfidf_vectorizer_terbaik.pkl' ada di direktori yang sama.")
    except Exception as e:
        print(f"❌ Terjadi error: {e}")
```
