```python
import pandas as pd
from google_play_scraper import reviews, Sort
from tqdm import tqdm
import time

# --- PENGATURAN UTAMA (Ubah sesuai kebutuhan) ---

# ID Aplikasi Mobile JKN di Google Play Store
APP_ID = "app.bpjs.mobile" 

# Jumlah review NETRAL yang ingin Anda kumpulkan
JUMLAH_REVIEW_TARGET = 50000 

# Bintang yang dianggap NETRAL (Gunakan angka 3)
BINTANG_FILTER = 3

# Nama file output untuk menyimpan hasil
NAMA_FILE_OUTPUT = "hasil_scraping_netral_baru.csv"

# --- AKHIR PENGATURAN ---


# Inisialisasi list untuk menampung hasil
hasil_reviews = []
# Token untuk melanjutkan scraping halaman berikutnya
continuation_token = None

print(f"🚀 Memulai scraping untuk {JUMLAH_REVIEW_TARGET} ulasan Bintang {BINTANG_FILTER} dari aplikasi '{APP_ID}'...")

# Progress bar untuk memantau proses
with tqdm(total=JUMLAH_REVIEW_TARGET, unit="ulasan") as pbar:
    while len(hasil_reviews) < JUMLAH_REVIEW_TARGET:
        try:
            # Melakukan request ke Google Play Store
            new_reviews, continuation_token = reviews(
                APP_ID,
                lang='id',                           # Bahasa: Indonesia
                country='id',                        # Negara: Indonesia
                sort=Sort.NEWEST,                    # Urutkan dari yang terbaru untuk mendapatkan data relevan
                count=3300,                           # Ambil 200 ulasan per permintaan (jumlah efisien)
                filter_score_with=BINTANG_FILTER,    # <-- INI KUNCI UTAMANYA: Filter hanya ulasan bintang 3
                continuation_token=continuation_token
            )

            # Jika tidak ada hasil baru atau token habis, hentikan loop
            if not new_reviews or continuation_token is None:
                print("\n⚠️ Tidak ada ulasan baru yang ditemukan atau sudah mencapai akhir halaman.")
                break
            
            # Tambahkan hasil baru ke list utama
            hasil_reviews.extend(new_reviews)
            # Update progress bar
            pbar.update(len(new_reviews))
            
            # Beri jeda sedikit antar request agar tidak dianggap spam
            time.sleep(0.5)

        except Exception as e:
            print(f"\n❌ Terjadi error: {e}. Menunggu 10 detik sebelum mencoba lagi...")
            time.sleep(10)


print(f"\n✅ Scraping selesai. Total ulasan terkumpul: {len(hasil_reviews)}")

# Jika ada data yang terkumpul, simpan ke CSV
if hasil_reviews:
    print(f"💾 Menyimpan data ke file '{NAMA_FILE_OUTPUT}'...")
    # Potong hasil jika melebihi target
    df = pd.DataFrame(hasil_reviews[:JUMLAH_REVIEW_TARGET])
    
    # Pilih kolom yang relevan saja agar file tidak terlalu besar
    df_final = df[['userName', 'content', 'score', 'at']]
    df_final.rename(columns={'content': 'Ulasan', 'score': 'Rating', 'at': 'Tanggal'}, inplace=True)
    
    df_final.to_csv(NAMA_FILE_OUTPUT, index=False)
    print("✅ Data berhasil disimpan.")
else:
    print("Tidak ada data untuk disimpan.")
```

```python
import pandas as pd

# --- PENGATURAN (Silakan diubah) ---

# Nama file dataset LAMA Anda (hasil scraping sebelumnya)
NAMA_FILE_LAMA = "hasil_output.csv"

# Nama file dataset NETRAL BARU Anda
NAMA_FILE_NETRAL = "dataset_final_netral_SIAP_DIGUNAKAN.csv"

# Nama file FINAL untuk menyimpan hasil akhir gabungan
NAMA_FILE_FINAL = "DATASET_LENGKAP_FINAL.csv"

# --- AKHIR PENGATURAN ---

try:
    # 1. Membaca kedua file CSV
    print(f"🔄 Membaca dataset lama: '{NAMA_FILE_LAMA}'...")
    df_lama = pd.read_csv(NAMA_FILE_LAMA)
    print(f"✅ Berhasil. Jumlah baris: {len(df_lama)}")

    print(f"🔄 Membaca dataset netral baru: '{NAMA_FILE_NETRAL}'...")
    df_netral = pd.read_csv(NAMA_FILE_NETRAL)
    print(f"✅ Berhasil. Jumlah baris: {len(df_netral)}")

    # 2. Menyamakan Nama Kolom (Langkah Krusial)
    # Kita buat semua nama kolom menjadi standar sebelum digabung.
    print("\n✨ Menyamakan nama kolom agar seragam...")
    
    kolom_mapping_standar = {
        'Ulasan': 'content', 'review': 'content', 'ulasan': 'content',
        'Rating': 'score', 'rating': 'score',
        'Tanggal': 'at', 'date': 'at',
        'userName': 'userName', 'Nama Pengguna': 'userName'
    }

    df_lama.rename(columns=kolom_mapping_standar, inplace=True)
    df_netral.rename(columns=kolom_mapping_standar, inplace=True)
    
    # 3. Menggabungkan kedua dataset
    print("\n🔗 Menggabungkan kedua dataset...")
    df_final = pd.concat([df_lama, df_netral], ignore_index=True)
    print(f"✅ Penggabungan berhasil. Total baris sebelum pembersihan: {len(df_final)}")

    # 4. Membersihkan Duplikat
    # Menghapus ulasan yang sama persis berdasarkan teksnya ('content')
    print("\n🧹 Menghapus data duplikat berdasarkan isi ulasan...")
    jumlah_sebelum = len(df_final)
    df_final.dropna(subset=['content'], inplace=True) # Hapus baris yg ulasannya kosong
    df_final.drop_duplicates(subset=['content'], keep='first', inplace=True)
    jumlah_sesudah = len(df_final)
    print(f"✅ Pembersihan duplikat selesai. {jumlah_sebelum - jumlah_sesudah} baris duplikat dihapus.")
    print(f"Total baris data unik: {jumlah_sesudah}")

    # 5. Mengurutkan Data
    print("\n📊 Mengonversi tanggal dan mengurutkan data dari 2025 ke 2020...")
    df_final['at'] = pd.to_datetime(df_final['at'], errors='coerce')
    df_final.dropna(subset=['at'], inplace=True) # Hapus baris yg tanggalnya tidak valid
    
    # Urutkan dari yang terbaru (descending)
    df_final.sort_values(by='at', ascending=False, inplace=True)
    print("✅ Data berhasil diurutkan.")

    # 6. Menyimpan hasil akhir
    print(f"\n💾 Menyimpan hasil akhir ke file '{NAMA_FILE_FINAL}'...")
    # Pilih kolom yang paling penting untuk disimpan
    kolom_pilihan = ['at', 'userName', 'score', 'content', 'Label']
    kolom_tersedia = [kolom for kolom in kolom_pilihan if kolom in df_final.columns]
    
    df_final[kolom_tersedia].to_csv(NAMA_FILE_FINAL, index=False)
    print("🎉🎉🎉 Selamat! Dataset final Anda telah siap digunakan!")

except FileNotFoundError as e:
    print(f"❌ ERROR: File tidak ditemukan. Pastikan nama file '{e.filename}' sudah benar.")
except Exception as e:
    print(f"❌ Terjadi sebuah error: {e}")
```

```python
# ==============================================================================
# TAHAP 1 (DITINGKATKAN): PREPROCESSING TEKS (Pembersihan & Normalisasi)
# ==============================================================================

# 1. IMPORT LIBRARY
import pandas as pd
import re
import time
from tqdm import tqdm

# ==============================================================================
# 2. KAMUS NORMALISASI (TETAP SAMA)
# ==============================================================================
REPLACEMENTS_DICT = {
    # --- Singkatan Umum & Typo ---
    'yg': 'yang', 'ga': 'tidak', 'gak': 'tidak', 'gk': 'tidak', 'enggak': 'tidak', 'nggak': 'tidak', 'ndak': 'tidak', 'ndk': 'tidak', 'gaada': 'tidak ada',
    'tdk': 'tidak', 'bkn': 'bukan', 'utk': 'untuk', 'dgn': 'dengan', 'krn': 'karena', 'karna': 'karena',
    'bgt': 'banget', 'jg': 'juga', 'aja': 'saja', 'jd': 'jadi', 'jgn': 'jangan', 'aza': 'saja',
    'klo': 'kalau', 'kalo': 'kalau', 'kl': 'kalau', 'knp': 'kenapa', 'gmn': 'bagaimana', 'gimana': 'bagaimana', 'bgmn': 'bagaimana', 'ginana': 'bagaimana',
    'sdh': 'sudah', 'udh': 'sudah', 'blm': 'belum', 'balum': 'belum', 'sblm': 'sebelum', 'blom': 'belum',
    'skrg': 'sekarang', 'skrng': 'sekarang', 'bbrp': 'beberapa', 'brp': 'berapa', 'byk': 'banyak', 'bnyk': 'banyak',
    'tp': 'tapi', 'tpi': 'tapi', 'pdhl': 'padahal',
    'trus': 'terus', 'trs': 'terus', 'bngt': 'banget', 'dpt': 'dapat', 'dr': 'dari',
    'kpd': 'kepada', 'kyk': 'seperti', 'ky': 'seperti', 'nih': 'ini', 'sih': 'sih', 'tuh': 'itu', 'ni': 'ini',
    'dong': 'dong', 'kok': 'kok', 'pake': 'pakai', 'dipake': 'dipakai', 'dipke': 'dipakai', 'dipakeitx': 'dipakai',
    'sm': 'sama', 'ama': 'sama', 'cmn': 'cuma',
    'kpn': 'kapan', 'tlg': 'tolong', 'tlong': 'tolong', 'mnt': 'minta', 'gpp': 'tidak apa-apa', 'donk': 'dong',
    'emg': 'memang', 'sbnrnya': 'sebenarnya', 'tq': 'terima kasih', 'yth': 'yang terhormat', 'sgt': 'sangat', 'stlh': 'setelah',
    'lbih': 'lebih', 'bner': 'benar', 'sgera': 'segera', 'slm': 'salam',
    'itx': 'itu', 'bknnya': 'bukannya', 'tmbh': 'tambah', 'mlh': 'malah',
    'pas': 'saat', 'kmrin': 'kemarin', 'i': 'ini', 'liat': 'lihat', 'libat': 'lihat',
    'masiah': 'masih', 'hancur': 'hancur', 'ancur': 'hancur', 'boro boro': 'boro-boro', 'boro2': 'boro-boro', 'gajelas': 'tidak jelas',
    'notif': 'notifikasi', 'didaftarin': 'didaftarkan', 'bwat': 'buat', 'komen': 'komentar', 'ngabisin': 'menghabiskan',
    'mala': 'malah', 'bredel': 'rusak', 'disaat': 'disaat', 'genting': 'genting', 'masukin': 'memasukkan', 'kocak': '[tertawa]',
    'bikin': 'membuat', 'muak': 'muak', 'sia': 'sia-sia', 'bosen': 'bosan',
    'aku': 'saya', 'aq': 'saya', 'gw': 'saya', 'gua': 'saya', 'gue': 'saya', 'w': 'saya', 'sy': 'saya', 'sya': 'saya',
    'kamu': 'kamu', 'lo': 'kamu', 'loe': 'kamu', 'lu': 'kamu', 'org': 'orang', 'kau': 'kamu',
    'x': 'kali', 'pe x': 'berkali kali', 'y': 'ya', 'maww': 'mau', 'mw': 'mau', 'af': 'maaf', 'maap': 'maaf', 'maunya': 'maunya',
    'wkwk': '[tertawa]', 'wkwkwk': '[tertawa]', 'hahaha': '[tertawa]', 'hehe': '[tertawa]', 'xixi': '[tertawa]',
    'hiks': '[sedih]', 'huhu': '[sedih]',
    'astaga': 'astaga', 'astaghfirullah': 'astaga', 'naudzubillah': 'naudzubillah',
    'anjim': '[umpatan]', 'anjir': '[umpatan]', 'anjay': 'umpatan', 'njir': '[umpatan]', 'jir': '[umpatan]',
    'opo': 'apa', 'iki': 'ini', 'iku': 'itu', 'kui': 'itu', 'piye': 'bagaimana', 'kepiye': 'bagaimana',
    'angel': 'sulit', 'abot': 'berat', 'rudet': 'rumit', 'suwe': 'lama', 'suwi': 'lama', 'seabad': 'lama sekali', 'lila': 'lama',
    'meneh': 'lagi', 'wae': 'saja', 'ae': 'saja', 'jan': 'sungguh', 'tenan': 'benar', 'tenan': 'sungguh',
    'ora': 'tidak', 'ra': 'tidak', 'raiso': 'tidak bisa', 'oraiso': 'tidak bisa', 'g bisa': 'tidak bisa', 'gak bisa': 'tidak bisa', 'gabisa': 'tidak bisa',
    'kabeh': 'semua', 'mbuh': 'tidak tahu', 'mumet': 'pusing', 'nang': 'di', 'neng': 'di',
    'jancok': '[umpatan]', 'matamu': '[umpatan]', 'cok': '[umpatan]', 'cuk': '[umpatan]', 'nanaon': 'apa-apaan', 'hese': 'sulit', 'kie': 'begini', 'ge': 'juga',
    'opo o': 'apa apa', 'wes': 'sudah', 'isoh': 'bisa', 'iso': 'bisa', 'bsa': 'bisa', 'bari jeung': 'tetapi', 'teu': 'tidak', 'cobaen': 'coba',
    'monggo': 'silakan', 'monggo': 'silakan', 'matur nuwun': 'terima kasih', 'suwun': 'terima kasih', 'asline': 'aslinya', 'nggwe': 'buat',
    'disek': 'dulu', 'sakdurunge': 'sebelumnya', 'bukaen': 'buka', 'koyok': 'seperti',
    'dongo': 'bodoh', 'dungu': 'bodoh', 'bego': 'bodoh', 'goblok': 'bodoh', 'gblk': 'bodoh', 'tolol': 'bodoh', 'tlol': 'tolol', 'dongok': 'bodoh',
    'anjing': '[umpatan]', 'anjiing': '[umpatan]', 'asu': '[umpatan]', 'hancok': '[umpatan]', 'anjeng': '[umpatan]',
    'bangsat': '[umpatan]', 'bgst': '[umpatan]', 'biadab': 'biadab',
    'kontol': '[umpatan]', 'memek': '[umpatan]', 'jembut': '[umpatan]', 'kntl': '[umpatan]',
    'tai': 'kotoran', 'taik': 'kotoran', 'taek': 'kotoran', 'kampret': '[umpatan]', 'sial': 'sialan', 'najis': '[umpatan]',
    'babi': '[umpatan]', 'binatang': '[umpatan]', 'bajingan': '[umpatan]', 'sampah': '[sampah]', 'bobrok': 'bobrok', 'kotoran': '[kotoran]', 'buwosok': 'busuk', 'bosok': 'busuk',
    'ngehek': '[umpatan]', 'entut': 'kentut', 'mungak': 'muak',
    'apk': 'aplikasi', 'app': 'aplikasi', 'apps': 'aplikasi', 'apliksi': 'aplikasi', 'uplikasinya': 'aplikasinya', 'aplikasinya': 'aplikasinya',
    'eror': 'error', 'erorr': 'error', 'erorrr': 'error', 'ngebug': 'bug',
    'lemot': 'lambat', 'lelet': 'lambat', 'loding': 'loading',
    'update': 'pembaruan', 'updet': 'pembaruan', 'diupdate': 'diperbarui', 'pembaruan': 'pembaruan', 'diperbaharui': 'diperbarui', 'abdet': 'pembaruan',
    'login': 'masuk', 'log in': 'masuk', 'logout': 'keluar', 'log out': 'keluar', 'msuk': 'masuk',
    'register': 'daftar', 'registrasi': 'pendaftaran', 'daptar': 'daftar', 'pndftran': 'pendaftaran', 'didaptar': 'didaftar', 'daftarin': 'daftarkan',
    'verif': 'verifikasi', 'ferivikasi': 'verifikasi', 'perivikasi': 'verifikasi', 'verifikasi': 'verifikasi', 'verivikasi': 'verifikasi', 'blel': 'gagal', 'captcha': 'captcha', 'captca': 'captcha',
    'faskes': 'fasilitas kesehatan', 'faskesnya': 'fasilitas kesehatannya',
    'bpjs': 'bpjs', 'jkn': 'jkn', 'kesehatan': 'kesehatan', 'pasien': 'pasien',
    'cs': 'customer service', 'admin': 'admin', 'operator': 'operator',
    'kartu digital': 'kartu digital', 'kartu fisik': 'kartu fisik', 'kk': 'kartu keluarga', 'ktp': 'kartu tanda penduduk', 'kluarga': 'keluarga',
    'tagihan': 'tagihan', 'iuran': 'iuran', 'premi': 'premi',
    'no hp': 'nomor ponsel', 'nomor hp': 'nomor ponsel', 'no peserta': 'nomor peserta', 'nomer': 'nomor',
    'password': 'password', 'pasword': 'password', 'pssword': 'password', 'asword': 'password',
    'cesar': 'sesar', 'operasi': 'operasi', 'autodebit': 'autodebit', 'nik': 'nik',
    'diklik': 'diklik', 'diklik': 'diklik', 'muncul': 'muncul', 'nungguan': 'menunggu', 'oflline': 'offline', 'offline': 'offline',
    'bulak balek': 'bolak-balik', 'bulak bali': 'bolak-balik',
    'bagusss': 'bagus', 'baguss': 'bagus', 'bgs': 'bagus', 'good': 'bagus', 'gud': 'bagus',
    'jelek': 'jelek', 'jlek': 'jelek', 'jellek': 'jelek', 'jelex': 'jelek', 'bapuk': 'jelek',
    'susah': 'sulit', 'ribet': 'rumit', 'gampang': 'mudah',
    'mantap': 'mantap', 'mntp': 'mantap', 'mantul': 'mantap betul',
    'ok': 'oke', 'okeey': 'oke',
    'makasih': 'terima kasih', 'mksh': 'terima kasih', 'thx': 'terima kasih', 'terimakasih': 'terima kasih',
    'tolongg': 'tolong', 'plis': 'tolong', 'pls': 'tolong',
    'respon': 'respons', 'responnya': 'responsnya',
    'bangettt': 'banget', 'bangett': 'banget', 'bgtt': 'banget',
    'parah': 'parah', 'males': 'malas', 'mager': 'malas gerak',
    'kecewa': 'kecewa', 'kcewa': 'kecewa',
    'bintang': 'bintang', 'bntg': 'bintang', 'nol': 'nol',
    'fungsi': 'fungsi', 'fungsik': 'fungsi', 'guna': 'guna',
    'nyampe': 'sampai', 'sampe': 'sampai', 'dikirim': 'dikirim', 'balesan': 'balasan',
    'jt': 'juta', 'rb': 'ribu', 'th': 'tahun', 'bln': 'bulan', 'wktu': 'waktu', 'ttap': 'tetap',
    'ketrangannya': 'keterangannya', 'pnjelasanya': 'penjelasannya', 'mngeluh': 'mengeluh', 'tambahkn': 'tambahkan', 'angak': 'angka',
    'kedepannya': 'kedepannya', 'dihapus': 'dihapus', 'playstore': 'playstore', 'google': 'google', 'macam': 'seperti',
    'terpakse': 'terpaksa', 'selanjutnya': 'selanjutnya', 'giliran': 'giliran', 'aktif': 'aktif', 'sayang': 'sayang', 'nya': 'nya', 'mikir': 'pikir',
    'miskin': 'miskin', 'rakyat': 'rakyat', 'gaji': 'gaji', 'naik': 'naik', 'boom': '[ekspresi]',
    'korup': 'korupsi', 'mulu': 'terus', 'idih': '[ekspresi]', 'sms': 'sms', 'pulsa': 'pulsa', 'min': 'admin',
    'becus': 'becus', 'klen': 'kalian', 'adeuh': '[ekspresi]', 'ambyar': 'hancur'
}

# ==============================================================================
# 3. FUNGSI UTAMA UNTUK PREPROCESSING (DITINGKATKAN)
# ==============================================================================
def preprocess_text(text, replacements):
    """
    Fungsi lengkap untuk membersihkan dan menormalisasi teks.
    """
    # Pastikan input adalah string
    text = str(text)
    
    # 1. Ubah ke huruf kecil
    text = text.lower()
    
    # 2. Hapus URL
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    
    # 3. Hapus angka
    text = re.sub(r'\d+', '', text)
    
    # 4. Terapkan kamus normalisasi (metode regex Anda yang efisien)
    regex = re.compile(r'\b(' + '|'.join(re.escape(key) for key in replacements.keys()) + r')\b')
    text = regex.sub(lambda match: replacements[match.group(0)], text)
    
    # 5. Hapus karakter spesial dan tanda baca (kecuali underscore untuk kata negasi)
    text = re.sub(r'[^a-zA-Z0-9\s_]', '', text)
    
    # 6. Hapus spasi berlebih
    text = ' '.join(text.split())
    
    return text

def run_preprocessing_pipeline(input_filename, output_filename, text_column='content'):
    """
    Alur kerja utama untuk tahap preprocessing.
    """
    try:
        start_time = time.time()
        print(f"▶️  Memulai Tahap 1: Preprocessing Teks untuk file '{input_filename}'...")
        
        # --- Memuat Data ---
        print(f"🔄 Memuat data dari '{input_filename}'...")
        df = pd.read_csv(input_filename)
        df.dropna(subset=[text_column], inplace=True)
        print(f"✅ Data berhasil dimuat. Jumlah baris: {len(df)}")
        
        # --- Menampilkan Pratinjau SEBELUM ---
        print("\n--- 🧐 Contoh 5 Ulasan SEBELUM diproses ---")
        print(df[text_column].head().to_string())
        
        # --- Menerapkan Preprocessing dengan Progress Bar ---
        print(f"\n🔄 Menerapkan pembersihan dan normalisasi pada {len(df)} baris...")
        # Daftarkan tqdm dengan pandas untuk progress_apply
        tqdm.pandas(desc="Processing Reviews")
        df['processed_text'] = df[text_column].progress_apply(lambda x: preprocess_text(x, REPLACEMENTS_DICT))
        
        # --- Menampilkan Pratinjau SESUDAH ---
        print("\n--- ✨ Contoh 5 Ulasan SESUDAH diproses ---")
        print(df[['processed_text', text_column]].head().to_string())
        
        # --- Menyimpan Hasil ---
        print(f"\n💾 Menyimpan hasil ke file '{output_filename}'...")
        df.to_csv(output_filename, index=False, encoding='utf-8-sig')
        
        end_time = time.time()
        print(f"\n✅ Tahap 1 Selesai. Total waktu: {int(end_time - start_time)} detik.")
        print(f"   - File '{output_filename}' telah dibuat dan berisi kolom 'processed_text'.")

    except FileNotFoundError:
        print(f"❌ Error: File '{input_filename}' tidak ditemukan.")
    except KeyError:
        print(f"❌ Error: Kolom '{text_column}' tidak ditemukan di file Anda.")
    except Exception as e:
        print(f"❌ Terjadi kesalahan tak terduga: {e}")

# ==============================================================================
# 4. JALANKAN PROSES
# ==============================================================================
if __name__ == "__main__":
    # --- KONFIGURASI FILE ---
    INPUT_FILE = "DATASET_SIAP_DILABELI.csv"  # Pastikan nama file ini benar
    OUTPUT_FILE = "1_hasil_preprocessing.csv"
    TEXT_COLUMN_NAME = "content" 
    
    run_preprocessing_pipeline(
        input_filename=INPUT_FILE, 
        output_filename=OUTPUT_FILE,
        text_column=TEXT_COLUMN_NAME
    )
```

```python
# ==============================================================================
# TAHAP 2 (VERSI AMAN): ANALISIS & LAPORAN USULAN PERUBAHAN LABEL
# ==============================================================================
import pandas as pd
import time

# --- PENGATURAN (Bisa diubah) ---
INPUT_FILE = "1_hasil_preprocessing.csv" # Gunakan file hasil preprocessing Tahap 1
OUTPUT_LAPORAN_FILE = "3_laporan_usulan_perubahan_label.csv"
TEXT_COLUMN = 'processed_text' # Kolom berisi teks bersih dari tahap sebelumnya
LABEL_COLUMN = 'Label'

# KUNCI KEAMANAN: Berapa banyak kata kunci yang harus ada sebelum label diusulkan berubah?
# Angka 2 lebih aman daripada 1.
AMBANG_BATAS_KEYWORD = 2 

# --- KAMUS KATA KUNCI (Sama seperti sebelumnya) ---
KATA_KUNCI_NEGATIF_KUAT = [
    'gagal', 'error', 'eror', 'sulit', 'lambat', 'kecewa', 'buruk', 'parah', 
    'rumit', 'masalah', 'tidak bisa', 'sampah', 'jelek', 'bodoh', 'mengecewakan', 
    'bobrok', 'hancur', 'susah', 'muak', 'sia-sia', 'rusak', 'sialan', '[umpatan]'
]
KATA_KUNCI_POSITIF_KUAT = [
    'terima kasih', 'mantap', 'terbantu', 'mudah', 'lancar', 'cepat', 
    'bagus sekali', 'luar biasa', 'sangat baik', 'sempurna', 'keren', 
    'bermanfaat', 'sangat membantu', 'sangat bagus'
]

def analyze_and_propose_changes(input_filename, output_filename, text_column, label_column, threshold):
    try:
        start_time = time.time()
        print(f"▶️  Memulai Analisis Usulan Perubahan Label untuk file '{input_filename}'...")
        
        df = pd.read_csv(input_filename)
        df.dropna(subset=[text_column], inplace=True)
        print(f"✅ Data dimuat. Jumlah baris: {len(df)}")
        
        print(f"🔄 Menganalisis teks dengan ambang batas {threshold} kata kunci...")

        # Membuat pola regex untuk menghitung kemunculan kata kunci
        pola_negatif = r'\b(' + '|'.join(KATA_KUNCI_NEGATIF_KUAT) + r')\b'
        pola_positif = r'\b(' + '|'.join(KATA_KUNCI_POSITIF_KUAT) + r')\b'

        df['jumlah_negatif'] = df[text_column].str.count(pola_negatif)
        df['jumlah_positif'] = df[text_column].str.count(pola_positif)

        # Kondisi untuk mengusulkan perubahan (dengan ambang batas)
        usul_ke_negatif = (df[label_column] == 'Netral') & (df['jumlah_negatif'] >= threshold) & (df['jumlah_positif'] == 0)
        usul_ke_positif = (df[label_column] == 'Netral') & (df['jumlah_positif'] >= threshold) & (df['jumlah_negatif'] == 0)
        
        # Gabungkan semua baris yang diusulkan untuk berubah
        df_usulan_negatif = df[usul_ke_negatif].copy()
        df_usulan_negatif['usulan_label_baru'] = 'Negatif'
        
        df_usulan_positif = df[usul_ke_positif].copy()
        df_usulan_positif['usulan_label_baru'] = 'Positif'
        
        df_laporan = pd.concat([df_usulan_negatif, df_usulan_positif])
        
        if not df_laporan.empty:
            print(f"✅ Analisis selesai. Ditemukan {len(df_laporan)} ulasan yang diusulkan untuk diubah labelnya.")
            
            # Buat kolom alasan untuk mempermudah review
            def cari_keyword(teks, keywords):
                ditemukan = [k for k in keywords if k in teks]
                return ', '.join(ditemukan)

            df_laporan['alasan_perubahan (keyword ditemukan)'] = df_laporan.apply(
                lambda row: cari_keyword(row[text_column], KATA_KUNCI_NEGATIF_KUAT) if row['usulan_label_baru'] == 'Negatif' 
                else cari_keyword(row[text_column], KATA_KUNCI_POSITIF_KUAT), axis=1
            )

            # Pilih kolom yang relevan untuk laporan
            kolom_laporan = ['at', 'score', 'content', text_column, label_column, 'usulan_label_baru', 'alasan_perubahan (keyword ditemukan)']
            df_laporan_final = df_laporan[[col for col in kolom_laporan if col in df_laporan.columns]]

            print(f"💾 Menyimpan laporan ke file '{output_filename}'...")
            df_laporan_final.to_csv(output_filename, index=False, encoding='utf-8-sig')
            print("🎉 Berhasil! Silakan buka dan review file laporan tersebut.")
            print("   Jika ada usulan yang tidak Anda setujui, cukup hapus barisnya dari file laporan.")
        else:
            print("✅ Analisis selesai. Tidak ada label yang perlu diubah berdasarkan aturan yang ditetapkan.")

    except Exception as e:
        print(f"❌ Terjadi kesalahan: {e}")

if __name__ == "__main__":
    analyze_and_propose_changes(
        input_filename=INPUT_FILE, 
        output_filename=OUTPUT_LAPORAN_FILE,
        text_column=TEXT_COLUMN,
        label_column=LABEL_COLUMN,
        threshold=AMBANG_BATAS_KEYWORD
    )
```

```python
# ==============================================================================
# TAHAP 2B: MENERAPKAN PERUBAHAN LABEL DARI LAPORAN
# ==============================================================================
import pandas as pd

# --- PENGATURAN ---
FILE_DATA_ASLI = "1_hasil_preprocessing.csv"
FILE_LAPORAN = "3_laporan_usulan_perubahan_label.csv"
FILE_OUTPUT_FINAL = "4_hasil_koreksi_label_final.csv"
TEXT_COLUMN = 'content' # Kolom unik untuk identifikasi baris

def apply_changes_from_report(original_file, report_file, output_file, key_column):
    try:
        print(f"🔄 Memuat data asli dari '{original_file}'...")
        df_asli = pd.read_csv(original_file)
        df_asli.set_index(key_column, inplace=True)

        print(f"🔄 Memuat laporan perubahan dari '{report_file}'...")
        df_laporan = pd.read_csv(report_file)
        df_laporan.set_index(key_column, inplace=True)
        
        # Ambil hanya label baru dari laporan
        perubahan_label = df_laporan['usulan_label_baru']
        
        print(f"🚀 Menerapkan {len(perubahan_label)} perubahan label...")
        # Update label di DataFrame asli berdasarkan indeks (teks ulasan)
        df_asli['Label'].update(perubahan_label)
        
        df_asli.reset_index(inplace=True)

        print(f"💾 Menyimpan hasil akhir ke '{output_file}'...")
        df_asli.to_csv(output_file, index=False, encoding='utf-8-sig')
        print("✅ Semua perubahan telah diterapkan. Dataset Anda telah diperbarui.")

    except FileNotFoundError:
        print(f"❌ Error: Pastikan file '{original_file}' dan '{report_file}' ada di folder yang sama.")
    except Exception as e:
        print(f"❌ Terjadi kesalahan: {e}")

if __name__ == "__main__":
    apply_changes_from_report(
        original_file=FILE_DATA_ASLI,
        report_file=FILE_LAPORAN,
        output_file=FILE_OUTPUT_FINAL,
        key_column=TEXT_COLUMN
    )
```

```python
# ==============================================================================
# SKRIP UNTUK TAHAP 2: KOREKSI LABEL
# ==============================================================================
import pandas as pd
import time

# --- PENGATURAN ---
INPUT_FILE = "1_hasil_preprocessing.csv"  # Hasil dari tahap normalisasi
OUTPUT_FILE = "2_hasil_koreksi_label.csv"
TEXT_COLUMN = 'processed_text'
LABEL_COLUMN = 'Label'

# --- Kamus Kata Kunci ---
KATA_KUNCI_NEGATIF_KUAT = [
    'gagal', 'error', 'eror', 'sulit', 'lambat', 'kecewa', 'buruk', 'parah', 
    'rumit', 'masalah', 'tidak bisa', 'sampah', 'jelek', 'bodoh', 'mengecewakan', 
    'bobrok', 'hancur', 'susah', 'muak', 'sia-sia', 'rusak', 'sialan', '[umpatan]'
]
KATA_KUNCI_POSITIF_KUAT = [
    'terima kasih', 'mantap', 'terbantu', 'mudah', 'lancar', 'cepat', 
    'bagus sekali', 'luar biasa', 'sangat baik', 'sempurna', 'keren', 
    'bermanfaat', 'sangat membantu', 'sangat bagus'
]

def correct_labels(df):
    pola_negatif = '|'.join(KATA_KUNCI_NEGATIF_KUAT)
    pola_positif = '|'.join(KATA_KUNCI_POSITIF_KUAT)
    df['ada_negatif'] = df[TEXT_COLUMN].str.contains(pola_negatif, na=False)
    df['ada_positif'] = df[TEXT_COLUMN].str.contains(pola_positif, na=False)
    
    kondisi_netral_ke_negatif = (df[LABEL_COLUMN] == 'Netral') & (df['ada_negatif']) & (~df['ada_positif'])
    df.loc[kondisi_netral_ke_negatif, LABEL_COLUMN] = 'Negatif'
    
    kondisi_netral_ke_positif = (df[LABEL_COLUMN] == 'Netral') & (df['ada_positif']) & (~df['ada_negatif'])
    df.loc[kondisi_netral_ke_positif, LABEL_COLUMN] = 'Positif'
    
    df.drop(columns=['ada_negatif', 'ada_positif'], inplace=True)
    return df

# --- Alur Kerja Utama ---
if __name__ == "__main__":
    print(f"▶️  Memulai Tahap 2: Koreksi Label untuk file '{INPUT_FILE}'...")
    df = pd.read_csv(INPUT_FILE)
    df.dropna(subset=[TEXT_COLUMN], inplace=True)
    
    print("\nDistribusi Label SEBELUM Koreksi:")
    print(df[LABEL_COLUMN].value_counts())
    
    df = correct_labels(df)
    
    print("\nDistribusi Label SESUDAH Koreksi:")
    print(df[LABEL_COLUMN].value_counts())
    
    df.to_csv(OUTPUT_FILE, index=False, encoding='utf-8-sig')
    print(f"\n✅ Tahap 2 Selesai. Hasil disimpan di '{OUTPUT_FILE}'.")
```

```python
# ==============================================================================
# TAHAP 5 (FINAL): STEMMING
# Tugas: Mengubah setiap kata dalam daftar token ke bentuk kata dasarnya.
# ==============================================================================

# 1. IMPORT LIBRARY
import pandas as pd
from ast import literal_eval # Untuk mengubah string list menjadi objek list
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import time
import sys

# ==============================================================================
# 2. FUNGSI UTAMA UNTUK STEMMING
# ==============================================================================

def run_stemming_pipeline(input_filename, output_filename, token_column='tokens_no_stopwords'):
    """
    Alur kerja utama untuk tahap stemming, sambil mempertahankan semua kolom.
    """
    try:
        start_time = time.time()
        print(f"▶️  Memulai Tahap 5 (Final): Stemming untuk file '{input_filename}'...")
        print("⚠️  PERINGATAN: Proses ini akan berjalan SANGAT LAMBAT. Harap bersabar.")
        
        # --- Memuat Data Hasil Stopword Removal ---
        print(f"🔄 Memuat data dari '{input_filename}'...")
        df = pd.read_csv(input_filename)
        df.dropna(subset=[token_column], inplace=True)
        print(f"✅ Data berhasil dimuat. Jumlah baris yang akan di-stem: {len(df)}")
        
        # --- Inisialisasi Stemmer Sastrawi ---
        print("🛠️  Menginisialisasi stemmer (Sastrawi)...")
        factory = StemmerFactory()
        stemmer = factory.create_stemmer()
        print("✅ Stemmer siap.")
        
        # --- Proses Stemming ---
        print(f"🔄 Melakukan stemming pada {len(df)} baris. Silakan tunggu...")
        
        # Mengubah kolom token yang tersimpan sebagai string kembali menjadi list
        df[token_column] = df[token_column].apply(literal_eval)
        
        stemmed_tokens_list = []
        total_rows = len(df)
        
        # Looping melalui setiap baris data
        for i, tokens in enumerate(df[token_column]):
            # Melakukan stemming pada setiap kata di dalam list token
            stemmed_tokens = [stemmer.stem(word) for word in tokens]
            stemmed_tokens_list.append(stemmed_tokens)
            
            # Memberikan laporan progres setiap 1000 baris
            if (i + 1) % 1000 == 0:
                elapsed_time = time.time() - start_time
                print(f"    ... stemming baris ke-{i+1} dari {total_rows} (waktu berjalan: {time.strftime('%M menit %S detik', time.gmtime(elapsed_time))})")
        
        # Membuat kolom baru dengan hasil stemming
        df['stemmed_tokens'] = stemmed_tokens_list
        
        # --- Menyimpan Hasil Final ---
        print("\n✅ Proses stemming selesai.")
        
        # Hapus baris yang mungkin menjadi kosong setelah stemming
        initial_rows = len(df)
        df = df[df['stemmed_tokens'].apply(len) > 0].reset_index(drop=True)
        if initial_rows > len(df):
            print(f"    - Info: Menghapus {initial_rows - len(df)} baris karena menjadi kosong setelah semua proses.")
        
        print(f"💾 Menyimpan hasil akhir pra-pemrosesan ke file '{output_filename}'...")
        df.to_csv(output_filename, index=False, encoding='utf-8-sig')
        
        end_time = time.time()
        total_time = end_time - start_time
        
        print(f"✅ Tahap 5 Selesai. Total waktu keseluruhan: {time.strftime('%H jam %M menit %S detik', time.gmtime(total_time))}.")
        print("\n" + "="*50)
        print("🎉 SELURUH PROSES PRA-PEMROSESAN DATA TELAH SELESAI!")
        print(f"🎯 File final Anda adalah '{output_filename}'.")
        print("="*50)

    except FileNotFoundError:
        print(f"❌ Error: File '{input_filename}' tidak ditemukan. Pastikan Anda sudah menjalankan Tahap 4.")
    except Exception as e:
        print(f"❌ Terjadi kesalahan tak terduga: {e}")

# ==============================================================================
# 3. JALANKAN PROSES
# ==============================================================================
if __name__ == "__main__":
    # --- KONFIGURASI FILE ---
    # Pastikan nama file ini adalah output dari Tahap 4
    INPUT_FILE = "4_hasil_stopwords_removal.csv"
    OUTPUT_FILE = "hasil_final_preprocessed.csv"
    
    run_stemming_pipeline(
        input_filename=INPUT_FILE, 
        output_filename=OUTPUT_FILE
    )
```
